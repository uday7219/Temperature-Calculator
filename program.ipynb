# 🌡️ Temperature Calculator

**Author:** Uday Sharma

A simple Python Temperature Calculator that converts temperature between:

- Celsius to Fahrenheit
- Celsius to Kelvin
- Fahrenheit to Celsius
- Fahrenheit to Kelvin
- Kelvin to Celsius
- Kelvin to Fahrenheit

This project demonstrates the use of:
- Functions
- Loops
- Conditional Statements
- User Input

```python
# Import required functions from the temperature module
from temperature import(
    show_menu,
    num,
    celsius_to_fahrenheit,
    celsius_to_kelvin,
    fahrenheit_to_celsius,
    fahrenheit_to_kelvin,
    kelvin_to_celsius,
    kelvin_to_fahrenheit
)
# Run the calculator continuously until the user exits
while True:
    # Display the conversion menu
    show_menu()
    # Take user's choice
    choice = input("Enter your choice : ")
    # Option 1: Celsius to Fahrenheit
    if choice == "1":
        temp = num()
        print(f"\nResult : {celsius_to_fahrenheit(temp):.2f} °F")
    # Option 2: Celsius to Kelvin
    elif choice == "2":
        temp = num()
        print(f"\nResult : {celsius_to_kelvin(temp):.2f} K")
    # Option 3: Fahrenheit to Celsius
    elif choice == "3":
        temp = num()
        print(f"\nResult : {fahrenheit_to_celsius(temp):.2f} °C")
    # Option 4: Fahrenheit to Kelvin
    elif choice == "4":
        temp = num()
        print(f"\nResult : {fahrenheit_to_kelvin(temp):.2f} K")
    # Option 5: Kelvin to Celsius
    elif choice == "5":
        temp = num()
        print(f"\nResult : {kelvin_to_celsius(temp):.2f} °C")
    # Option 6: Kelvin to Fahrenheit
    elif choice == "6":
        temp = num()
        print(f"\nResult : {kelvin_to_fahrenheit(temp):.2f} °F")
    # Option 7: Exit
    elif choice == "7":
        print("\nThank You for using Temperature Calculator ❤️")
        break
    # Display an error for invalid input
    else:
        print("\nInvalid Choice.")
```
