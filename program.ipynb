# 🌡️ Temperature Calculator

**Author:** Uday Sharma

A simple Python Temperature Calculator that converts temperature between:

- Celsius to Fahrenheit
- Celsius to Kelvin
- Fahrenheit to Celsius
- Fahrenheit to Kelvin
- Kelvin to Celsius
- Kelvin to Fahrenheit

This project demonstrates the use of:
- Functions
- Loops
- Conditional Statements
- User Input

```python
from temperature import(
    show_menu,
    num,
    celsius_to_fahrenheit,
    celsius_to_kelvin,
    fahrenheit_to_celsius,
    fahrenheit_to_kelvin,
    kelvin_to_celsius,
    kelvin_to_fahrenheit
)
while True:
    show_menu()
    choice = input("Enter your choice : ")

    if choice == "1":
        temp = num()
        print(f"\nResult : {celsius_to_fahrenheit(temp):.2f} °F")
    elif choice == "2":
        temp = num()
        print(f"\nResult : {celsius_to_kelvin(temp):.2f} K")
    elif choice == "3":
        temp = num()
        print(f"\nResult : {fahrenheit_to_celsius(temp):.2f} °C")
    elif choice == "4":
        temp = num()
        print(f"\nResult : {fahrenheit_to_kelvin(temp):.2f} K")
    elif choice == "5":
        temp = num()
        print(f"\nResult : {kelvin_to_celsius(temp):.2f} °C")
    elif choice == "6":
        temp = num()
        print(f"\nResult : {kelvin_to_fahrenheit(temp):.2f} °F")
    elif choice == "7":
        print("\nThank You for using Temperature Calculator ❤️")
        break
    else:
        print("\nInvalid Choice.")
```
